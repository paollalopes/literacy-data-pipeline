## Conclusões da Análise Exploratória

### Chaves e granularidade

- **avaliacao_alfabetizacao_municipio:** granularidade por município, ano e rede. Possível chave composta: `ano + id_municipio + rede`. A série é sempre 2 na fonte analisada.
- **avaliacao_alfabetizacao_uf:** granularidade por UF, ano e rede. Possível chave composta: `ano + sigla_uf + rede`. A série também é sempre 2.
- **meta_alfabetizacao_brasil:** granularidade nacional por ano e rede. Possível chave: `ano + rede`.
- **meta_alfabetizacao_municipio:** granularidade por município, ano e rede. Possível chave composta: `ano + id_municipio + rede`.
- **meta_alfabetizacao_uf:** granularidade por UF, ano e rede. Possível chave composta: `ano + sigla_uf + rede`.

As chaves foram consideradas possíveis chaves primárias porque não apresentaram repetições nos dados analisados. A unicidade deverá ser validada novamente durante o processamento da camada Silver.

### Relacionamentos identificados

Os principais relacionamentos identificados são:

- `avaliacao_alfabetizacao_municipio` ↔ `meta_alfabetizacao_municipio`
  - Chave territorial: `id_municipio`
  - Também devem ser considerados `ano` e a compatibilidade do domínio de `rede`.

- `avaliacao_alfabetizacao_uf` ↔ `meta_alfabetizacao_uf`
  - Chave territorial: `sigla_uf`
  - Também devem ser considerados `ano` e a compatibilidade do domínio de `rede`.

- `meta_alfabetizacao_brasil` possui granularidade nacional e poderá ser utilizada como referência para comparações agregadas, sem relacionamento direto por município ou UF.

### Problemas e características identificados

- Não foram encontrados registros completamente duplicados nas cinco fontes analisadas.
- Na avaliação municipal, as colunas `proporcao_aluno_nivel_0` até `proporcao_aluno_nivel_8` possuem valores nulos nos registros de 2023 e estão preenchidas em 2024.
- Na avaliação por UF ocorre comportamento semelhante nas colunas de proporção por nível.
- A avaliação por UF contém 25 UFs distintas, enquanto a base de metas por UF contém 27, indicando diferença de cobertura que deverá ser validada antes das integrações.
- A base de metas municipais possui valores ausentes em `taxa_alfabetizacao`, `nivel_alfabetizacao`, `percentual_participacao` e em metas de determinados períodos.
- `nivel_alfabetizacao` foi interpretado como `float64` devido à existência de valores nulos e deverá ter seu tipo semântico avaliado.
- A base de metas por UF também possui valores ausentes em indicadores e metas.
- Alguns campos são constantes nas fontes analisadas, como `serie = 2` nas avaliações e `meta_alfabetizacao_2030 = 80` nas metas preenchidas. Esses valores serão preservados, pois não há evidência na exploração de que representem erros.
